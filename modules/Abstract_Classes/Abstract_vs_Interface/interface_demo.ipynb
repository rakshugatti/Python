```python
#(a) Interface using Abstract Class
from abc import ABC, abstractmethod

# Interface-like abstract class
class PaymentInterface(ABC):

    @abstractmethod
    def pay(self):
        pass


# Implementation
class GooglePay(PaymentInterface):

    def pay(self):
        print("Payment done using Google Pay")


# Usage
gpay = GooglePay()
gpay.pay()
```

```python
#(b) Interface using Protocols(Python 3.8+)
from typing import Protocol

# Protocol interface
class PaymentProtocol(Protocol):
    def pay(self) -> None:
        pass


# No inheritance needed
class PhonePe:
    def pay(self):
        print("Payment done using PhonePe")


# Function using protocol
def make_payment(method: PaymentProtocol):
    method.pay()


# Usage
p = PhonePe()
make_payment(p)
```
